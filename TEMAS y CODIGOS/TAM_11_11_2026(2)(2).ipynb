# COSENO

```
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# Frecuencia de la señal (Hz)
f = 2
# Frecuencia de muestreo (Hz)
fs = 1000

# Vector de tiempo x basado en la frecuencia de muestreo
x = np.arange(-2, 2, 1/fs).reshape(-1, 1)

# Modelo de salida (Coseno)
y = np.cos(2 * np.pi * f * x)

plt.figure(figsize=(10, 4))
plt.plot(x, y)
plt.title('Señal Coseno')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.grid(True)
plt.show()
```

tomando la se.al del coseno anteriormente generada parte de los datos usando tran test split 80/20 generada adicionalmente una grafica 1 fila, 2 columnas ubicada la se.al dividida

```
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# Realizamos la partición de la señal coseno (80% train, 20% test)
x_train_cos, x_test_cos, y_train_cos, y_test_cos = train_test_split(x, y, test_size=0.20, random_state=42)


#
# Ordenamos los datos para que la visualización de la línea sea continua
sort_idx_train = np.argsort(x_train_cos.flatten())
sort_idx_test = np.argsort(x_test_cos.flatten())

# Creamos la gráfica de 1 fila y 2 columnas
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Gráfica del conjunto de entrenamiento
axes[0].scatter(x_train_cos, y_train_cos, color='blue', alpha=0.5, label='Datos Train')
axes[0].plot(x_train_cos.flatten()[sort_idx_train], y_train_cos.flatten()[sort_idx_train], color='darkblue', linestyle='--')
axes[0].set_title('Conjunto de Entrenamiento (80%)')
axes[0].set_xlabel('Tiempo (s)')
axes[0].set_ylabel('Amplitud')
axes[0].grid(True)
axes[0].legend()

# Gráfica del conjunto de prueba
axes[1].scatter(x_test_cos, y_test_cos, color='red', alpha=0.5, label='Datos Test')
axes[1].plot(x_test_cos.flatten()[sort_idx_test], y_test_cos.flatten()[sort_idx_test], color='darkred', linestyle='--')
axes[1].set_title('Conjunto de Prueba (20%)')
axes[1].set_xlabel('Tiempo (s)')
axes[1].set_ylabel('Amplitud')
axes[1].grid(True)
axes[1].legend()

plt.tight_layout()
plt.show()
print("tamaño x_train_cos",x_train_cos.shape)

print("tamaño x_test_cos",x_test_cos.shape)
```

```
from sklearn.model_selection import train_test_split

# Realizamos la partición 80/20
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)

# Mostramos las dimensiones resultantes
print(f"Tamaño de x_train: {x_train.shape}")
print(f"Tamaño de x_test: {x_test.shape}")

#MODELO
input = tf.keras.layers.Input(shape=(1,))
d1 = tf.keras.layers.Dense(64,activation="selu")(input)
d2 = tf.keras.layers.Dense(64,activation="selu")(d1)
d3 = tf.keras.layers.Dense(64,activation="selu")(d2)
output = tf.keras.layers.Dense(1)(d3)

Model = tf.keras.Model(inputs=input, outputs=output)
Model.compile(optimizer='adam', loss='mse', metrics=["mae"])
Model.summary()

#Compilar el modelo
Model.compile(optimizer='adam', loss='mse',metrics=["mae"]) #adam:actuliza pesos, mse:Error cuadrático medio,
```

```
Model.fit(x_train, y_train, epochs=400, batch_size=32)#Entrenamiento del modelo
```

```
plt.scatter(x_test_cos,y_test_cos)
plt.scatter(x_test_cos,Model.predict(x_test_cos))
```
