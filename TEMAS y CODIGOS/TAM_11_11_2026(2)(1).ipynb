```
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import time

# 1. Preparar las dimensiones correctas de prueba con tamaño 'w'
# Creamos un x_test que tenga la misma longitud de ventana (w=100) para poder evaluar adecuadamente los modelos.
x_test_w = []
limite = y.shape[0] - w - p - 1
for i in range(0, limite, p):
    # Usamos la ventana anterior a la predicción real para evaluar de manera justa
    x_test_w.append(y[i + p : i + p + w])

x_test_w = np.array(x_test_w)

# Formatos secuenciales 3D (batch, w, features) para RNN, LSTM y GRU
x_train_seq = np.expand_dims(x_train, axis=-1)
x_test_w_seq = np.expand_dims(x_test_w, axis=-1)

print("Shape x_train (DNN):", x_train.shape)
print("Shape x_test_w (DNN):", x_test_w.shape)

# 2. Definición de Modelos
def build_dnn():
    In = tf.keras.layers.Input(shape=(w,))
    d1 = tf.keras.layers.Dense(64, activation='relu')(In)
    d2 = tf.keras.layers.Dense(32, activation='relu')(d1)
    Out = tf.keras.layers.Dense(1)(d2)
    return tf.keras.Model(inputs=In, outputs=Out, name='DNN')

def build_rnn():
    In = tf.keras.layers.Input(shape=(w, 1))
    r1 = tf.keras.layers.SimpleRNN(32, activation='tanh')(In)
    Out = tf.keras.layers.Dense(1)(r1)
    return tf.keras.Model(inputs=In, outputs=Out, name='RNN')

def build_lstm():
    In = tf.keras.layers.Input(shape=(w, 1))
    l1 = tf.keras.layers.LSTM(32)(In)
    Out = tf.keras.layers.Dense(1)(l1)
    return tf.keras.Model(inputs=In, outputs=Out, name='LSTM')

def build_gru():
    In = tf.keras.layers.Input(shape=(w, 1))
    g1 = tf.keras.layers.GRU(32)(In)
    Out = tf.keras.layers.Dense(1)(g1)
    return tf.keras.Model(inputs=In, outputs=Out, name='GRU')

# Instanciar y compilar modelos con sus respectivos conjuntos de datos de tamaño compatible (w = 100)
models = {
    'DNN': (build_dnn(), x_train, x_test_w),
    'RNN': (build_rnn(), x_train_seq, x_test_w_seq),
    'LSTM': (build_lstm(), x_train_seq, x_test_w_seq),
    'GRU': (build_gru(), x_train_seq, x_test_w_seq)
}

results = {}

# Entrenar y evaluar
for name, (model, xtr, xte) in models.items():
    print(f"\n--- Entrenando {name} ---")
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    t0 = time.time()
    history = model.fit(xtr, y_train, epochs=15, batch_size=32, verbose=0, validation_split=0.1)
    t1 = time.time()

    loss, mae = model.evaluate(xte, y_test, verbose=0)
    results[name] = {'Loss (MSE)': loss, 'MAE': mae, 'Time (s)': t1 - t0}
    print(f"{name} -> MSE: {loss:.6f} | MAE: {mae:.6f} | Tiempo: {t1-t0:.2f}s")
```

```
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import time

# 1. Definición de la nueva serie de tiempo (Tendencia + Estacionalidad + Ruido)
np.random.seed(42)
t_seq = np.arange(0, 40, 0.01) # 4000 puntos equivalentes
trend = 0.02 * t_seq
seasonality = np.sin(2 * np.pi * 0.5 * t_seq) + 0.5 * np.cos(2 * np.pi * 1.2 * t_seq)
noise = np.random.normal(0, 0.1, size=t_seq.shape)
y_seq = trend + seasonality + noise

# 2. Preparar ventanas temporales (w=100, p=20)
w = 100
p = 20
limite_seq = len(y_seq) - w - p - 1

x_train_ts = []
y_train_ts = []
x_test_ts = []
y_test_ts = []

for i in range(0, limite_seq, p):
    x_train_ts.append(y_seq[i : i + w])
    y_train_ts.append(y_seq[i + w : i + w + 1])
    # Ventana de evaluación alineada para test
    x_test_ts.append(y_seq[i + p : i + p + w])
    y_test_ts.append(y_seq[i + w + p : i + w + p + 1])

x_train_ts = np.array(x_train_ts)
y_train_ts = np.array(y_train_ts)
x_test_ts = np.array(x_test_ts)
y_test_ts = np.array(y_test_ts)

# Formatear conjuntos de datos para redes recurrentes 3D (batch, timesteps, features)
x_train_ts_3d = np.expand_dims(x_train_ts, axis=-1)
x_test_ts_3d = np.expand_dims(x_test_ts, axis=-1)

# 3. Estructuración y entrenamiento rápido de los modelos
def get_dnn():
    In = tf.keras.layers.Input(shape=(w,))
    d1 = tf.keras.layers.Dense(64, activation='relu')(In)
    d2 = tf.keras.layers.Dense(32, activation='relu')(d1)
    Out = tf.keras.layers.Dense(1)(d2)
    return tf.keras.Model(inputs=In, outputs=Out)

def get_rnn():
    In = tf.keras.layers.Input(shape=(w, 1))
    r1 = tf.keras.layers.SimpleRNN(32, activation='tanh')(In)
    Out = tf.keras.layers.Dense(1)(r1)
    return tf.keras.Model(inputs=In, outputs=Out)

def get_lstm():
    In = tf.keras.layers.Input(shape=(w, 1))
    l1 = tf.keras.layers.LSTM(32)(In)
    Out = tf.keras.layers.Dense(1)(l1)
    return tf.keras.Model(inputs=In, outputs=Out)

def get_gru():
    In = tf.keras.layers.Input(shape=(w, 1))
    g1 = tf.keras.layers.GRU(32)(In)
    Out = tf.keras.layers.Dense(1)(g1)
    return tf.keras.Model(inputs=In, outputs=Out)

ts_models = {
    'DNN': (get_dnn(), x_train_ts, x_test_ts),
    'RNN': (get_rnn(), x_train_ts_3d, x_test_ts_3d),
    'LSTM': (get_lstm(), x_train_ts_3d, x_test_ts_3d),
    'GRU': (get_gru(), x_train_ts_3d, x_test_ts_3d)
}

# Ajustar pesos
for name, (model, xtr, xte) in ts_models.items():
    model.compile(optimizer='adam', loss='mse')
    model.fit(xtr, y_train_ts, epochs=15, batch_size=32, verbose=0)

# 4. Inferencia paso a paso (One-step-ahead)
pred_dnn = ts_models['DNN'][0].predict(x_test_ts, verbose=0).flatten()
pred_rnn = ts_models['RNN'][0].predict(x_test_ts_3d, verbose=0).flatten()
pred_lstm = ts_models['LSTM'][0].predict(x_test_ts_3d, verbose=0).flatten()
pred_gru = ts_models['GRU'][0].predict(x_test_ts_3d, verbose=0).flatten()

# 5. Predicción Recursiva Autorregresiva (50 pasos)
pasos_futuros = 50
ultima_ventana_ts = x_test_ts[-1].copy()

def predecir_recursivo_ts(model_name, keras_model):
    ventana = ultima_ventana_ts.copy()
    predicciones = []
    for _ in range(pasos_futuros):
        if model_name == 'DNN':
            pred = keras_model.predict(ventana.reshape(1, w), verbose=0)[0, 0]
        else:
            pred = keras_model.predict(ventana.reshape(1, w, 1), verbose=0)[0, 0]
        predicciones.append(pred)
        ventana = np.append(ventana[1:], pred)
    return predicciones

auto_dnn = predecir_recursivo_ts('DNN', ts_models['DNN'][0])
auto_rnn = predecir_recursivo_ts('RNN', ts_models['RNN'][0])
auto_lstm = predecir_recursivo_ts('LSTM', ts_models['LSTM'][0])
auto_gru = predecir_recursivo_ts('GRU', ts_models['GRU'][0])

# 6. Graficar resultados
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

t_test_ts = t_seq[w + p : w + p + len(x_test_ts) * p : p]

# Subplot 1: Inferencia sobre la nueva dinámica
axes[0].plot(t_seq[:1000], y_seq[:1000], label='Serie Temporal Real (Fondo 1000 pts)', color='#cccccc', linestyle='--')
axes[0].plot(t_test_ts[:40], y_test_ts[:40], label='Real Test', color='black', marker='o', linewidth=2, markersize=5)
axes[0].plot(t_test_ts[:40], pred_dnn[:40], label='DNN Predict', linestyle='-')
axes[0].plot(t_test_ts[:40], pred_rnn[:40], label='RNN Predict', linestyle='-')
axes[0].plot(t_test_ts[:40], pred_lstm[:40], label='LSTM Predict', linestyle='-')
axes[0].plot(t_test_ts[:40], pred_gru[:40], label='GRU Predict', linestyle='-')
axes[0].set_title('Inferencia de Dinámica en Serie Temporal Compleja (Test)')
axes[0].set_xlabel('Tiempo')
axes[0].set_ylabel('Valor')
axes[0].legend()
axes[0].grid(True)

# Subplot 2: Predicción Recursiva Autorregresiva
t_futuro = np.arange(pasos_futuros)
axes[1].plot(t_futuro, auto_dnn, label='DNN (Recursivo)', marker='x')
axes[1].plot(t_futuro, auto_rnn, label='RNN (Recursivo)', marker='o')
axes[1].plot(t_futuro, auto_lstm, label='LSTM (Recursivo)', marker='^')
axes[1].plot(t_futuro, auto_gru, label='GRU (Recursivo)', marker='s')
axes[1].set_title('Inferencia Autorregresiva con Ruido y Tendencia (50 Pasos)')
axes[1].set_xlabel('Pasos de Predicción hacia el futuro')
axes[1].set_ylabel('Valor Predicho')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()
```
